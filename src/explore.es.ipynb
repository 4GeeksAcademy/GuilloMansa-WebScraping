```python
import os
from bs4 import BeautifulSoup
import requests
import time
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

import requests 

import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL de la página
url = 'https://www.promiedos.com.ar/league/conmebol-libertadores/bac/historial'

response = requests.get(url)
response.status_code

response.content



```

```python
soup = BeautifulSoup(response.content)
soup
```

```python
champions = soup.find_all('tr', class_ = 'table-row row_row__vWDV9')
champions[0]
```

```python
len(champions)
```

```python
team_list = []

for trophy in champions[1:]:
    filters = trophy.find_all('td')
    if len(filters) >= 2:
        team_name = filters[0].get_text(strip=True)
        year = filters[1].get_text(strip=True)
        team_list.append((team_name, year))

team_list
```

```python
team_list = []

for trophy in champions[1:]:
    filters = trophy.find_all('td')
    if len(filters) >= 2:
        year_text = filters[0].get_text(strip=True)
        if year_text.isdigit():
            team_dict = {'Year': year_text,'Champion of this year': filters[1].get_text(strip=True)}
            team_list.append(team_dict)

team_list

```
